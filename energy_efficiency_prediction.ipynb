##Step 2: Data Collection

###Import Necessary Libraries

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor,AdaBoostRegressor,GradientBoostingRegressor
import xgboost as xgb
import lightgbm as lgbm
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score
import warnings
warnings.filterwarnings('ignore')
```

###Load Dataset

```python
energy_efficient_data = pd.read_csv('energy_efficiency_data.csv')
energy_efficient_data
```

##Step 3: Data Understanding

###Perform Initial Investigation

```python
energy_efficient_data.shape
```

```python
energy_efficient_data.isna().sum()
```

##Step 4: Data Preparation

```python
X = energy_efficient_data.drop(columns=['Heating_Load','Cooling_Load'])
X
```

```python
y = energy_efficient_data['Heating_Load']
y
```

```python
std_scaler = StandardScaler()
X_scaled = std_scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled,columns=X.columns)
X_scaled
```

##Step 5: Model Building

###Train Test Split

```python
X_train,X_test,y_train,y_test = train_test_split(X_scaled,y,test_size=0.2,random_state=13,shuffle=True)
```

```python
rf_model = RandomForestRegressor(n_estimators=50,max_depth=10)
```

```python
ada_boost = AdaBoostRegressor(n_estimators=50)
```

```python
grad_boost = GradientBoostingRegressor(n_estimators=50,learning_rate=0.1)
```

```python
xgrad_boost = xgb.XGBRegressor(n_estimators=50,max_depth=5,learning_rate=0.1)
```

```python
lgrad_boost = lgbm.LGBMRegressor(n_estimators=50,num_leaves=5,learning_rate=0.1,max_depth=5)
```

##Step 6: Model Training

```python
rf_model.fit(X_train,y_train)
```

```python
ada_boost.fit(X_train,y_train)
```

```python
grad_boost.fit(X_train,y_train)
```

```python
xgrad_boost.fit(X_train,y_train)
```

```python
lgrad_boost.fit(X_train,y_train)
```

##Step 7: Model Testing

###Training data prediction

```python
y_rf_model_train = rf_model.predict(X_train)
```

```python
y_ada_boost_train = ada_boost.predict(X_train)
```

```python
y_grad_boost_train = grad_boost.predict(X_train)
```

```python
y_xgrad_boost_train = xgrad_boost.predict(X_train)
```

```python
y_lgrad_boost_train = lgrad_boost.predict(X_train)
```

###Test data prediction

```python
y_rf_model_test = rf_model.predict(X_test)
```

```python
y_ada_boost_test = ada_boost.predict(X_test)
```

```python
y_grad_boost_test = grad_boost.predict(X_test)
```

```python
y_xgrad_boost_test = xgrad_boost.predict(X_test)
```

```python
y_lgrad_boost_test = lgrad_boost.predict(X_test)
```

##Step 8: Model Evaluation

```python
def adjusted_r2(r2,n,p):
  return 1-((1 - r2) *(n - 1) / (n - p - 1))
```

###Training Data Evaluation

```python
print('MAE:', round(mean_absolute_error(y_train,y_rf_model_train),4))
print('MSE:',round(mean_squared_error(y_train,y_rf_model_train),4))
print('RMSE:',round(root_mean_squared_error(y_train,y_rf_model_train),4))
print('R2 Score:',round(r2_score(y_train,y_rf_model_train),4))
rf_train_adj_r2 = adjusted_r2(
    r2_score(y_train,y_rf_model_train),
    X_train.shape[0],
    X_train.shape[1]
)
print('Adjusted R2 Score:',round(rf_train_adj_r2,4))
```
